```python
import pandas as pd

df = pd.read_csv("/content/01 covid_toy - 01 covid_toy.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
x = df.iloc[:, 0:5]
y = df.drop(columns=x.columns)
print(x.head())

print()
print()
print()

print(y.head())
```

```python
# Train-Test split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
print(x_train.head())
print()
print()
print(y_train.head())
```

```python
# Performing SimpleImputer on fever column for filling missing values

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder

si = SimpleImputer()
le = LabelEncoder()

# Fit and transform the imputer on training feature 'fever' and transform test feature 'fever'
x_train['fever'] = si.fit_transform(x_train[['fever']])
x_test['fever'] = si.transform(x_test[['fever']])

# Applying label encoding on the target 'has_covid' column
y_train['has_covid'] = le.fit_transform(y_train['has_covid'])
y_test['has_covid'] = le.transform(y_test['has_covid'])

df
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()

x_train['cough'] = oe.fit_transform(x_train[['cough']])
x_test['cough'] = oe.transform(x_test[['cough']])

x_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False)

x_train_gender_city = ohe.fit_transform(x_train[['gender', 'city']])
x_test_gender_city = ohe.transform(x_test[['gender', 'city']])

print(x_train_gender_city.shape)
print(x_test_gender_city.shape)
```
